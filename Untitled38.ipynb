```python
from __future__ import annotations

from collections import Counter
from pathlib import Path

from PIL import Image


# ============================================================
# Configuration
# ============================================================

DATASET_DIR = Path(
    r"C:\PATH\TO\HDEV6\training_data\extracted_imgs"
)

IMAGE_EXTENSIONS = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"}
METADATA_EXTENSIONS = {".csv", ".xlsx", ".xls"}


# ============================================================
# Dataset discovery
# ============================================================

def discover_files(root: Path) -> list[Path]:
    """Return all files under the dataset directory."""
    if not root.exists():
        raise FileNotFoundError(
            f"Dataset directory does not exist: {root}"
        )

    if not root.is_dir():
        raise NotADirectoryError(
            f"Dataset path is not a directory: {root}"
        )

    return [
        path
        for path in root.rglob("*")
        if path.is_file()
    ]


def classify_files(
    files: list[Path],
) -> tuple[list[Path], list[Path], list[Path]]:
    """
    Split discovered files into:
    - original images
    - shifted images
    - metadata files
    """
    original_images: list[Path] = []
    shifted_images: list[Path] = []
    metadata_files: list[Path] = []

    for path in files:
        suffix = path.suffix.lower()

        if suffix in IMAGE_EXTENSIONS:
            if "_shifted" in path.stem.lower():
                shifted_images.append(path)
            else:
                original_images.append(path)

        elif suffix in METADATA_EXTENSIONS:
            metadata_files.append(path)

    return original_images, shifted_images, metadata_files


# ============================================================
# Image validation
# ============================================================

def inspect_image_sizes(
    image_paths: list[Path],
) -> tuple[Counter, list[Path]]:
    """
    Read image dimensions without modifying the images.

    Returns:
        size_counts:
            Frequency of each (width, height).
        corrupted_images:
            Images that could not be opened/verified.
    """
    size_counts: Counter = Counter()
    corrupted_images: list[Path] = []

    for image_path in image_paths:
        try:
            with Image.open(image_path) as image:
                image.verify()

            with Image.open(image_path) as image:
                size_counts[image.size] += 1

        except Exception:
            corrupted_images.append(image_path)

    return size_counts, corrupted_images


# ============================================================
# Reporting
# ============================================================

def print_report(
    root: Path,
    all_files: list[Path],
    original_images: list[Path],
    shifted_images: list[Path],
    metadata_files: list[Path],
    size_counts: Counter,
    corrupted_images: list[Path],
) -> None:

    print("\n" + "=" * 60)
    print("DATASET INVENTORY")
    print("=" * 60)

    print(f"\nDataset:")
    print(f"  {root}")

    print("\nFiles:")
    print(f"  Total files:       {len(all_files)}")
    print(f"  Original images:   {len(original_images)}")
    print(f"  Shifted images:    {len(shifted_images)}")
    print(f"  Metadata files:    {len(metadata_files)}")

    print("\nImage dimensions:")
    for (width, height), count in size_counts.most_common():
        print(f"  {width} x {height}: {count}")

    print("\nValidation:")
    print(f"  Corrupted images:  {len(corrupted_images)}")

    if corrupted_images:
        print("\nCorrupted image files:")
        for path in corrupted_images[:20]:
            print(f"  - {path}")

        if len(corrupted_images) > 20:
            print(
                f"  ... and "
                f"{len(corrupted_images) - 20} more."
            )

    print("\n" + "=" * 60)


# ============================================================
# Entry point
# ============================================================

def main() -> None:
    files = discover_files(DATASET_DIR)

    (
        original_images,
        shifted_images,
        metadata_files,
    ) = classify_files(files)

    size_counts, corrupted_images = inspect_image_sizes(
        original_images
    )

    print_report(
        root=DATASET_DIR,
        all_files=files,
        original_images=original_images,
        shifted_images=shifted_images,
        metadata_files=metadata_files,
        size_counts=size_counts,
        corrupted_images=corrupted_images,
    )


if __name__ == "__main__":
    main()
```
